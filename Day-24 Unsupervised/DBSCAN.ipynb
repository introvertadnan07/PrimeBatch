```python
import pandas as pd
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import warnings
warnings.filterwarnings("ignore")
```

```python
iris = load_iris()
```

```python
iris
```

```python
X = iris.data
y = iris.target
```

```python
X.shape
```

```python
# Visualize
sns.scatterplot(x=X[:,0], y=X[:,2], c=y)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)
```

```python
x_scaled
```

```python
# Optional - dimensionlity reduction using PCA
```

```python
from sklearn.cluster import KMeans

wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)  # Fit the model on your dataset
    wcss.append(kmeans.inertia_)  # Append to list
```

```python
sns.lineplot(x=range(1, 11), y=wcss, marker='o')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # create X_scaled
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X[:,0], y=X[:,2], c=labels)
```
